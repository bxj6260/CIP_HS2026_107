## Data explorer for ACLED data
file renamed to `ACLED_data_yemen.csv` and  `ACLED_data_mid_east.csv` for simplification. Explorer uses folium for map import.

```python
import pandas as pd
import folium
from narwhals import DataFrame
```

```python
data = pd.read_csv("ACLED_data_mid_east.csv")
#data = data[data["admin1"] == "North Indian Ocean"]
data
```

```python
unique_locations= data[['centroid_latitude', "centroid_longitude", 'country', 'event_type']].drop_duplicates()
unique_locations
```

```python
data['country'].value_counts()
```

```python
avg_lat = data["centroid_latitude"].mean()
avg_long = data['centroid_longitude'].mean()
center =[avg_lat, avg_long]


# Create interactive map
m = folium.Map(location=center, zoom_start=5)
```

```python
for i, v in unique_locations.iterrows():
    loc = [v["centroid_latitude"], v["centroid_longitude"]]
    folium.Marker(location=loc,
                  popup=v[['country', 'event_type']]
                  ).add_to(m)

m
```
